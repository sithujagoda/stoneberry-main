```python
from pathlib import Path

DATASET_PATH = Path("../datasets/raw/gemstones-images-expanded/dataset-expanded")

print(DATASET_PATH.exists())
```

```python
classes = sorted([folder.name for folder in DATASET_PATH.iterdir() if folder.is_dir()])

print(f"Total Gem Types: {len(classes)}")
print(classes[:10])
```

```python
from pathlib import Path

counts = {}

for folder in DATASET_PATH.iterdir():
    if folder.is_dir():
        counts[folder.name] = len(list(folder.glob("*.jpg")))

print(counts)
```

```python
import pandas as pd

df = pd.DataFrame(
    counts.items(),
    columns=["Gemstone", "Images"]
)

df.sort_values("Images", ascending=False).head()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(18,6))
plt.bar(df["Gemstone"], df["Images"])
plt.xticks(rotation=90)
plt.title("Number of Images per Gemstone")
plt.show()
```

```python
import matplotlib.pyplot as plt
from PIL import Image
import random

gem = random.choice(classes)

image = random.choice(list((DATASET_PATH / gem).glob("*.jpg")))

img = Image.open(image)

plt.figure(figsize=(5,5))
plt.imshow(img)
plt.title(gem)
plt.axis("off")
plt.show()
```
