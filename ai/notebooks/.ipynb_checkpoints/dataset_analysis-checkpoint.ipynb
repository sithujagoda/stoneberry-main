```python
from pathlib import Path

TRAIN = Path("../datasets/processed/train")
VAL = Path("../datasets/processed/validation")
TEST = Path("../datasets/processed/test")
```

```python
def count_images(folder):
    return len(list(folder.glob("*")))

for split in [TRAIN, VAL, TEST]:

    print(f"\n{split.name.upper()}")

    print("Gemstone :", count_images(split / "Gemstone"))
    print("NotGem   :", count_images(split / "NotGem"))
```

```python
import random
import matplotlib.pyplot as plt
from PIL import Image

def show_random_images(folder, title):

    images = random.sample(list(folder.glob("*")), 9)

    plt.figure(figsize=(10,10))

    for i, img_path in enumerate(images):

        plt.subplot(3,3,i+1)

        img = Image.open(img_path)

        plt.imshow(img)

        plt.axis("off")

    plt.suptitle(title)

    plt.show()
```

```python
show_random_images(TRAIN/"Gemstone", "Gemstone")
```

```python
show_random_images(TRAIN/"NotGem", "NotGem")
```
